```python
SCENARIO: “Hospital Smart Assistant System”
🏥 Background Story
A large hospital deploys an AI-powered patient assistant.
👉 Patients can ask:
- “What is my appointment schedule?”
- “What are my latest test results?”
👉 Instead of calling reception or logging into multiple portals,
👉 AI fetches it instantly, providing secure, real-time updates.
```

```python
import datetime

MOCK_EHR_DB = {
    "Patient/99": {
        "name": "John Doe",
        "appointments": [
            {"id": "a1", "date": "2026-05-10T14:30:00", "dept": "Radiology", "status": "confirmed"},
            {"id": "a2", "date": "2026-06-01T09:00:00", "dept": "General Practice", "status": "pending"}
        ],
        "observations": [
            {"code": "Glucose", "value": "98 mg/dL", "date": "2026-03-25", "sensitive": False},
            {"code": "Biopsy", "value": "Critical: Results Pending", "date": "2026-03-29", "sensitive": True}
        ]
    }
}

class HospitalAssistant:
    def __init__(self, patient_id):
        self.patient_id = f"Patient/{patient_id}"
        self.data = MOCK_EHR_DB.get(self.patient_id)

    def get_appointments(self):
        if not self.data: return "Patient not found."

        apps = self.data['appointments']
        response = "📅 **Upcoming Appointments:**\n"
        for a in apps:
            dt = datetime.datetime.fromisoformat(a['date'])
            response += f"- {dt.strftime('%b %d at %I:%M %p')} ({a['dept']})\n"
        return response

    def get_lab_results(self):
        if not self.data: return "Patient not found."

        results = self.data['observations']
        response = "**Latest Test Results:**\n"
        for r in results:
            if r['sensitive']:
                response += f"- {r['code']}: [Information restricted. Please contact your physician directly.]\n"
            else:
                response += f"- {r['code']}: {r['value']} (Tested on {r['date']})\n"
        return response

# --- EXECUTION LOOP ---
def run_assistant():
    # Simulated Authenticated Session for Patient ID 99
    assistant = HospitalAssistant("99")

    print("Hospital System: Verified Identity for John Doe.\n")

    # Query 1: Schedule
    print(assistant.get_appointments())

    # Query 2: Results
    print(assistant.get_lab_results())

if __name__ == "__main__":
    run_assistant()
```

SCENARIO: “Banking Smart Assistant System”
🏦 Background Story
A major bank launches an AI-powered customer assistant.
👉 Customers can ask:
- “What is my account balance?”
- “Show me my last 5 transactions.”
- “When is my loan EMI due?”
👉 Instead of logging into apps or waiting on customer service calls,

⚙️ Core Idea:
Just like the hospital and college scenarios, the assistant removes manual checking, centralizes financial data, and makes access instant.
💡 Impact:

- Reduces load on call centers.
- Provides personalized financial insights on demand.

```python
# ======================================
# STEP 1: Install Libraries
# ======================================
!pip install groq gradio

# ======================================
# STEP 2: Load API Key
# ======================================
from google.colab import userdata

groq_api_key = userdata.get("api_key")

if not groq_api_key:
    raise ValueError("❌ GROQ_API_KEY not found")

from groq import Groq
client = Groq(api_key=groq_api_key)

# ======================================
# STEP 3: Dummy Bank Database
# ======================================
accounts = {
    "1001": {
        "name": "Ravi",
        "balance": 50000,
        "transactions": [
            "₹2000 - Grocery",
            "₹5000 - Rent",
            "₹800 - Fuel",
            "₹1200 - Food",
            "₹3000 - Shopping"
        ],
        "emi_due": "5 April 2026"
    },
    "1002": {
        "name": "Neha",
        "balance": 75000,
        "transactions": [
            "₹1000 - Cab",
            "₹2500 - Shopping",
            "₹700 - Food",
            "₹4000 - Rent",
            "₹1500 - Bills"
        ],
        "emi_due": "10 April 2026"
    }
}

# ======================================
# STEP 4: Tool Functions
# ======================================
def get_balance(account_id):
    if account_id in accounts:
        return f"💰 Balance: ₹{accounts[account_id]['balance']}"
    return "❌ Account not found"

def get_transactions(account_id):
    if account_id in accounts:
        txns = "\n".join(accounts[account_id]['transactions'])
        return f"📄 Last 5 Transactions:\n{txns}"
    return "❌ Account not found"

def get_emi(account_id):
    if account_id in accounts:
        return f"🏦 EMI Due Date: {accounts[account_id]['emi_due']}"
    return "❌ Account not found"

# ======================================
# STEP 5: RBAC Security
# ======================================
def secure_access(role, user_account, requested_account, action):

    if role == "manager":
        return True

    elif role == "employee":
        # Employees cannot see EMI or approve loans
        if action == "get_emi":
            return False
        return True

    elif role == "customer":
        return user_account == requested_account

    return False

# ======================================
# STEP 6: MCP Decision via LLM
# ======================================
def decide_action(query):
    try:
        prompt = f"""
        You are a banking assistant.

        Choose one action:
        - get_balance
        - get_transactions
        - get_emi

        Rules:
        - Balance → get_balance
        - Transactions → get_transactions
        - EMI / loan due → get_emi

        Return ONLY action name.

        Query: {query}
        """

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": prompt}]
        )

        return response.choices[0].message.content.strip().lower()

    except Exception as e:
        print("❌ Error:", e)
        return "fallback"

# ======================================
# STEP 7: Agent Logic
# ======================================
def banking_agent(message, role, user_account, requested_account, history):

    if not user_account:
        response = "⚠️ Enter your account ID"
        history.append((message, response))
        return "", history

    if not requested_account:
        requested_account = user_account

    # Step 1: Decide action
    action = decide_action(message)

    # Step 2: RBAC Check
    if not secure_access(role, user_account, requested_account, action):
        response = "🚫 Access Denied"

    else:
        # Step 3: Tool Execution
        if "balance" in action:
            response = get_balance(requested_account)

        elif "transaction" in action:
            response = get_transactions(requested_account)

        elif "emi" in action:
            response = get_emi(requested_account)

        # fallback
        elif action == "fallback":
            msg = message.lower()
            if "balance" in msg:
                response = get_balance(requested_account)
            elif "transaction" in msg:
                response = get_transactions(requested_account)
            elif "emi" in msg or "loan" in msg:
                response = get_emi(requested_account)
            else:
                response = "⚠️ Could not understand"

        else:
            response = "🤖 Ask about balance, transactions, or EMI"

    history.append((message, response))
    return "", history

# ======================================
# STEP 8: Gradio UI
# ======================================
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown("# 🏦 Banking Smart Assistant (MCP + RBAC + Groq)")

    role = gr.Dropdown(
        ["customer", "employee", "manager"],
        label="Select Role"
    )

    user_account = gr.Textbox(label="Your Account ID (e.g., 1001)")
    requested_account = gr.Textbox(label="Target Account ID (optional)")

    chatbot = gr.Chatbot(height=400)
    msg = gr.Textbox(label="Ask your question")

    state = gr.State([])

    msg.submit(
        banking_agent,
        inputs=[msg, role, user_account, requested_account, state],
        outputs=[msg, chatbot]
    )

# ======================================
# STEP 9: Launch App
# ======================================
demo.launch(share=True)
```

```python
# ======================================
# 🚀 UNIVERSITY SMART ASSISTANT (ONE CELL)
# ======================================

!pip install groq gradio -q

# ======================================
# 🔑 API KEY (Set your Groq key here)
# ======================================
from google.colab import userdata
api_key = userdata.get('api_key')

from groq import Groq
client = Groq(api_key=api_key)

# ======================================
# 🏫 DUMMY UNIVERSITY DATABASE
# ======================================
students = {
    "S101": {
        "name": "Arnav",
        "attendance": "85%",
        "results": "CGPA: 8.7"
    },
    "S102": {
        "name": "Riya",
        "attendance": "92%",
        "results": "CGPA: 9.1"
    }
}

courses = {
    "CSE101": {
        "schedule": "Mon-Wed 10AM",
        "pending_approvals": ["S102"]
    }
}

# ======================================
# 🛠️ TOOL FUNCTIONS
# ======================================
def get_attendance(student_id):
    if student_id in students:
        return f"📊 Attendance: {students[student_id]['attendance']}"
    return "❌ Student not found"

def get_results(student_id):
    if student_id in students:
        return f"📄 Results: {students[student_id]['results']}"
    return "❌ Student not found"

def update_schedule(course_id):
    if course_id in courses:
        courses[course_id]["schedule"] = "Updated: Tue-Thu 2PM"
        return f"✅ Schedule updated for {course_id}"
    return "❌ Course not found"

def approve_registration(course_id):
    if course_id in courses:
        courses[course_id]["pending_approvals"] = []
        return f"✅ Registrations approved for {course_id}"
    return "❌ Course not found"

# ======================================
# 🔐 RBAC SECURITY
# ======================================
def secure_access(role, user_id, action):

    if role == "admin":
        return True

    elif role == "faculty":
        if action in ["update_schedule", "approve_registration"]:
            return True
        return False

    elif role == "student":
        if action in ["get_attendance", "get_results"]:
            return True
        return False

    return False

# ======================================
# 🤖 MCP DECISION (LLM)
# ======================================
def decide_action(query):
    try:
        prompt = f"""
        You are a university assistant.

        Choose one action:
        - get_attendance
        - get_results
        - update_schedule
        - approve_registration

        Rules:
        - Attendance → get_attendance
        - Results → get_results
        - Update schedule → update_schedule
        - Approve registrations → approve_registration

        Return ONLY action name.

        Query: {query}
        """

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": prompt}]
        )

        return response.choices[0].message.content.strip().lower()

    except:
        return "fallback"

# ======================================
# 🧠 AGENT LOGIC
# ======================================
def university_agent(message, role, user_id, history):

    if not user_id:
        response = "⚠️ Enter your ID"
        history.append((message, response))
        return "", history

    action = decide_action(message)

    # 🔐 RBAC CHECK
    if not secure_access(role, user_id, action):
        response = "🚫 Access Denied"

    else:
        # 🛠️ TOOL EXECUTION
        if "attendance" in action:
            response = get_attendance(user_id)

        elif "result" in action:
            response = get_results(user_id)

        elif "schedule" in action:
            response = update_schedule("CSE101")

        elif "approve" in action:
            response = approve_registration("CSE101")

        # fallback
        elif action == "fallback":
            msg = message.lower()
            if "attendance" in msg:
                response = get_attendance(user_id)
            elif "result" in msg:
                response = get_results(user_id)
            elif "schedule" in msg:
                response = update_schedule("CSE101")
            elif "approve" in msg:
                response = approve_registration("CSE101")
            else:
                response = "⚠️ Could not understand"

        else:
            response = "🤖 Ask about attendance, results, schedule, or approvals"

    history.append((message, response))
    return "", history

# ======================================
# 🎨 GRADIO UI
# ======================================
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown("# 🎓 University Smart Assistant (MCP + RBAC)")

    role = gr.Dropdown(
        ["student", "faculty", "admin"],
        label="Select Role"
    )

    user_id = gr.Textbox(label="Enter ID (e.g., S101)")

    chatbot = gr.Chatbot(height=400)
    msg = gr.Textbox(label="Ask your question")

    state = gr.State([])

    msg.submit(
        university_agent,
        inputs=[msg, role, user_id, state],
        outputs=[msg, chatbot]
    )

# ======================================
# 🚀 RUN APP
# ======================================
demo.launch(share=True)
```

```python
# ======================================
# 🏬 RETAIL SMART ASSISTANT (ONE CELL)
# ======================================

!pip install groq gradio -q

# ======================================
# 🔑 API KEY
# ======================================

from groq import Groq
client = Groq(api_key= 'api_key')

# ======================================
# 🏬 DUMMY DATABASE
# ======================================
customers = {
    "C101": {
        "name": "Arnav",
        "purchase_history": [
            "₹2000 - Shoes",
            "₹1500 - T-shirt",
            "₹3000 - Headphones"
        ]
    },
    "C102": {
        "name": "Riya",
        "purchase_history": [
            "₹1000 - Bag",
            "₹2500 - Watch"
        ]
    }
}

inventory = {
    "shoes": 25,
    "t-shirt": 50,
    "headphones": 10
}

supplier_orders = ["Order#5001", "Order#5002"]

employee_schedule = {
    "emp1": "Mon-Wed Morning",
    "emp2": "Thu-Fri Evening"
}

# ======================================
# 🛠️ TOOL FUNCTIONS
# ======================================
def get_purchase_history(customer_id):
    if customer_id in customers:
        return "🛒 Purchase History:\n" + "\n".join(customers[customer_id]["purchase_history"])
    return "❌ Customer not found"

def check_inventory(product):
    product = product.lower()
    if product in inventory:
        return f"📦 {product} stock: {inventory[product]}"
    return "❌ Product not found"

def approve_supplier_orders():
    global supplier_orders
    supplier_orders = []
    return "✅ All supplier orders approved"

def manage_schedule():
    employee_schedule["emp1"] = "Updated: Weekend Shift"
    return "📅 Employee schedule updated"

# ======================================
# 🔐 RBAC SECURITY
# ======================================
def secure_access(role, user_id, action):

    if role == "admin":
        return True

    elif role == "manager":
        if action in ["check_inventory", "approve_orders", "manage_schedule"]:
            return True
        return False

    elif role == "customer":
        if action == "get_purchase":
            return True
        return False

    return False

# ======================================
# 🤖 MCP DECISION (LLM)
# ======================================
def decide_action(query):
    try:
        prompt = f"""
        You are a retail assistant.

        Choose one action:
        - get_purchase
        - check_inventory
        - approve_orders
        - manage_schedule

        Rules:
        - Purchase history → get_purchase
        - Inventory → check_inventory
        - Approve supplier → approve_orders
        - Schedule → manage_schedule

        Return ONLY action name.

        Query: {query}
        """

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": prompt}]
        )

        return response.choices[0].message.content.strip().lower()

    except:
        return "fallback"

# ======================================
# 🧠 AGENT LOGIC
# ======================================
def retail_agent(message, role, user_id, history):

    if not user_id:
        response = "⚠️ Enter your ID"
        history.append((message, response))
        return "", history

    action = decide_action(message)

    # 🔐 RBAC CHECK
    if not secure_access(role, user_id, action):
        response = "🚫 Access Denied"

    else:
        # 🛠️ TOOL EXECUTION
        if "purchase" in action:
            response = get_purchase_history(user_id)

        elif "inventory" in action:
            # extract product (simple way)
            product = message.split()[-1]
            response = check_inventory(product)

        elif "approve" in action:
            response = approve_supplier_orders()

        elif "schedule" in action:
            response = manage_schedule()

        # fallback
        elif action == "fallback":
            msg = message.lower()
            if "purchase" in msg:
                response = get_purchase_history(user_id)
            elif "inventory" in msg:
                product = msg.split()[-1]
                response = check_inventory(product)
            elif "approve" in msg:
                response = approve_supplier_orders()
            elif "schedule" in msg:
                response = manage_schedule()
            else:
                response = "⚠️ Could not understand"

        else:
            response = "🤖 Ask about purchases, inventory, orders, or schedules"

    history.append((message, response))
    return "", history

# ======================================
# 🎨 GRADIO UI
# ======================================
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown("# 🏬 Retail Smart Assistant (MCP + RBAC)")

    role = gr.Dropdown(
        ["customer", "manager", "admin"],
        label="Select Role"
    )

    user_id = gr.Textbox(label="Enter ID (e.g., C101)")

    chatbot = gr.Chatbot(height=400)
    msg = gr.Textbox(label="Ask your question")

    state = gr.State([])

    msg.submit(
        retail_agent,
        inputs=[msg, role, user_id, state],
        outputs=[msg, chatbot]
    )

# ======================================
# 🚀 RUN
# ======================================
demo.launch(share=True)
```

```python
# SCENARIO: “AI Banking Assistant with Role-Based Access”
# 🏦 Background Story

# A bank builds an AI assistant to help users:

# Check account balance
# View transactions
# Approve loans
# Manage customer accounts

# 👉 But not everyone can do everything

# 🧑‍🤝‍🧑 Roles in the Bank
# Role	Description
# 👤 Customer	Bank account holder
# 👨‍💼 Employee	Bank staff
# 🧑‍💼 Manager	Branch manager
# 🔐 Permissions (RBAC)
# Action	Customer	Employee	Manager
# View own balance	✅	✅	✅
# View others' accounts	❌	✅	✅
# Approve loan	❌	❌	✅
# View all transactions	❌	✅	✅


# ======================================
# STEP 1: Install Libraries
# ======================================
!pip install groq gradio


# ======================================
# STEP 2: Load API Key from Colab Secrets
# ======================================
from google.colab import userdata

groq_api_key = userdata.get("GROQ_API_KEY")

if not groq_api_key:
    raise ValueError("❌ GROQ_API_KEY not found in Colab Secrets")

from groq import Groq
client = Groq(api_key=groq_api_key)


# ======================================
# STEP 3: Dummy Bank Database
# ======================================
accounts = {
    "1001": {"name": "Amit", "balance": 50000},
    "1002": {"name": "Neha", "balance": 75000}
}


# ======================================
# STEP 4: Tool Functions (APIs)
# ======================================
def get_balance(account_id):
    if account_id in accounts:
        return f"💰 Balance of {account_id}: ₹{accounts[account_id]['balance']}"
    return "❌ Account not found"


def approve_loan(account_id):
    if account_id in accounts:
        return f"🏦 Loan approved for account {account_id}"
    return "❌ Account not found"


# ======================================
# STEP 5: RBAC Security Layer
# ======================================
def secure_access(role, user_account, requested_account, action):

    # Manager → full access
    if role == "manager":
        return True

    # Employee → can view all but cannot approve loans
    elif role == "employee":
        if action == "approve_loan":
            return False
        return True

    # Customer → only own account, no loan approval
    elif role == "customer":
        return user_account == requested_account and action != "approve_loan"

    return False


# ======================================
# STEP 6: MCP Tool Decision via LLM
# ======================================
def decide_action(query):
    try:
        prompt = f"""
        You are an AI banking assistant.

        Decide which action to take:
        - get_balance
        - approve_loan

        Rules:
        - Balance related queries → get_balance
        - Loan approval queries → approve_loan

        Return ONLY the action name.

        Query: {query}
        """

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": prompt}]
        )

        action = response.choices[0].message.content.strip().lower()
        return action

    except Exception as e:
        print("❌ Groq Error:", e)
        return "fallback"


# ======================================
# STEP 7: MCP Agent (Core Logic)
# ======================================
def banking_agent(message, role, user_account, requested_account, history):

    # Input validation
    if not user_account:
        response = "⚠️ Please enter your account ID"
        history.append((message, response))
        return "", history

    if not requested_account:
        requested_account = user_account  # default to own account

    # Step 1: LLM decides action
    action = decide_action(message)

    # Step 2: RBAC Security Check
    if not secure_access(role, user_account, requested_account, action):
        response = "🚫 Access Denied: You are not authorized"

    else:
        # Step 3: Tool Invocation
        if "balance" in action:
            response = get_balance(requested_account)

        elif "loan" in action:
            response = approve_loan(requested_account)

        # Fallback if LLM fails
        elif action == "fallback":
            msg_lower = message.lower()
            if "balance" in msg_lower:
                response = get_balance(requested_account)
            elif "loan" in msg_lower:
                response = approve_loan(requested_account)
            else:
                response = "⚠️ Could not understand request"

        else:
            response = "🤖 Try asking about balance or loan"

    # Save chat history
    history.append((message, response))

    return "", history


# ======================================
# STEP 8: Gradio UI
# ======================================
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown("# 🏦 AI Banking Assistant (MCP + RBAC + Groq)")

    role = gr.Dropdown(
        ["customer", "employee", "manager"],
        label="Select Role"
    )

    user_account = gr.Textbox(label="Your Account ID (e.g., 1001)")
    requested_account = gr.Textbox(label="Target Account ID (optional)")

    chatbot = gr.Chatbot(height=400)
    msg = gr.Textbox(label="Ask your question")

    state = gr.State([])

    msg.submit(
        banking_agent,
        inputs=[msg, role, user_account, requested_account, state],
        outputs=[msg, chatbot]
    )


# ======================================
# STEP 9: Launch App
# ======================================
demo.launch(share=True)
```

SCENARIO: “AI Healthcare Support Assistant in a Large Hospital”
🏢 Background Story
A large hospital (like Apollo or Fortis) has thousands of patients and staff members.
👉 Daily Challenges Faced:
- Patients waiting hours for appointment confirmations
- Confusion about lab test results and reports
- Doctors overwhelmed with scheduling and follow-up reminders
- Nurses struggling to track medicine administration times
- Emergency cases needing instant triage
👉 Instead of manual coordination, the hospital builds an:
🤖 AI Healthcare Support Assistant
🚑 Capabilities:
- 📅 Smart Scheduling: Automatically books and reschedules patient appointments based on doctor availability.
- 🧪 Lab Report Explainer: Summarizes test results in simple language for patients.
- 💊 Medication Tracker: Sends reminders to nurses and patients about dosage timings.
- 🚨 Emergency Triage: Instantly categorizes incoming cases (critical, urgent, routine) and alerts the right medical team.
- 📧 Follow-up Automation: Sends personalized recovery instructions and reminders after discharge.
🎯 Impact:
- Reduced patient waiting time
- Doctors spend more time on treatment, less on admin work
- Nurses avoid errors in medicine administration
- Faster response in emergencies
- Improved patient satisfaction and trust

```python
# ======================================
# 🏥 AI HEALTHCARE SUPPORT ASSISTANT
# ======================================

!pip install groq gradio -q

# ======================================
# 🔑 API KEY
# ======================================


from groq import Groq
client = Groq(api_key='api_key')

import time

# ======================================
# 🗂️ DUMMY DATABASE
# ======================================
appointments = []
medications = {
    "P101": ["8AM - Paracetamol", "2PM - Antibiotic", "8PM - Vitamin"]
}

# ======================================
# 🛠️ TOOL FUNCTIONS
# ======================================

# 📅 Scheduling
def book_appointment(patient_id):
    appointments.append(patient_id)
    return f"📅 Appointment booked for {patient_id}"

# 🧪 Lab Report Explainer (LLM)
def explain_report(report_text):
    prompt = f"""
    Explain this medical report in simple terms:

    {report_text}
    """
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

# 💊 Medication Tracker
def get_medications(patient_id):
    if patient_id in medications:
        return "💊 Medication Schedule:\n" + "\n".join(medications[patient_id])
    return "❌ No medication data"

# 🚨 Emergency Triage
def triage(symptoms):
    symptoms = symptoms.lower()
    if "chest pain" in symptoms or "unconscious" in symptoms:
        return "🚨 CRITICAL: Immediate ICU attention required"
    elif "fever" in symptoms or "pain" in symptoms:
        return "⚠️ URGENT: Doctor consultation needed"
    else:
        return "🟢 ROUTINE: Regular check-up"

# 📧 Follow-up
def follow_up(patient_id):
    return f"📧 Follow-up: Stay hydrated, take medicines on time, revisit after 7 days ({patient_id})"

# ======================================
# 🤖 MCP DECISION (LLM)
# ======================================
def decide_action(query):
    try:
        prompt = f"""
        You are a hospital assistant.

        Choose one action:
        - schedule
        - report
        - medication
        - triage
        - followup

        Rules:
        - Appointment → schedule
        - Report → report
        - Medicine → medication
        - Emergency → triage
        - Follow-up → followup

        Return ONLY action name.

        Query: {query}
        """

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": prompt}]
        )

        return response.choices[0].message.content.strip().lower()

    except:
        return "fallback"

# ======================================
# 🧠 AGENT LOGIC
# ======================================
def hospital_agent(message, patient_id, history):

    if not patient_id:
        response = "⚠️ Enter patient ID"
        history.append((message, response))
        return "", history

    action = decide_action(message)

    # 🛠️ TOOL EXECUTION
    if "schedule" in action:
        response = book_appointment(patient_id)

    elif "report" in action:
        response = explain_report(message)

    elif "medication" in action:
        response = get_medications(patient_id)

    elif "triage" in action:
        response = triage(message)

    elif "follow" in action:
        response = follow_up(patient_id)

    # fallback
    elif action == "fallback":
        msg = message.lower()
        if "appointment" in msg:
            response = book_appointment(patient_id)
        elif "report" in msg:
            response = explain_report(message)
        elif "medicine" in msg:
            response = get_medications(patient_id)
        elif "pain" in msg or "fever" in msg:
            response = triage(message)
        else:
            response = "⚠️ Could not understand"

    else:
        response = "🤖 Ask about appointment, report, medication, emergency, or follow-up"

    history.append((message, response))
    return "", history

# ======================================
# 🎨 GRADIO UI
# ======================================
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown("# 🏥 AI Healthcare Support Assistant")

    patient_id = gr.Textbox(label="Enter Patient ID (e.g., P101)")

    chatbot = gr.Chatbot(height=500)
    msg = gr.Textbox(label="Ask your question")

    state = gr.State([])

    msg.submit(
        hospital_agent,
        inputs=[msg, patient_id, state],
        outputs=[msg, chatbot]
    )

# ======================================
# 🚀 RUN
# ======================================
demo.launch(share=True)
```
