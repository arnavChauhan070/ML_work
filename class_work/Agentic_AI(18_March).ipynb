```python
# Scenario-Based Teaching: Smart Travel Agent
# 🧑‍🏫 How You Start in Class

# Say this:

# “Imagine you are very busy and you tell your assistant:

# ‘Find me the best flight from Delhi to Mumbai.’

# Now instead of replying with just text, your assistant actually thinks,
# checks data, and gives a final decision.”


# Step 1 — Define Tools

# 👉  “Tools = functions the agent can call”

def search_flights(source, destination):
    return [
        {"airline": "IndiGo", "price": 4500, "time": "6:00 AM"},
        {"airline": "Air India", "price": 5200, "time": "9:00 AM"}
    ]

def check_weather(city):
    return {"city": city, "temperature": 28, "condition": "Clear"}

def calculate_budget(flights):
    cheapest = min(flights, key=lambda x: x["price"])
    return cheapest
```

```python
# Step 2 — Agent Brain (Decision + Planning)
def agent_brain(user_query):
    plan = []

    if "flight" in user_query.lower():
        plan.append("search_flights")
        plan.append("check_weather")
        plan.append("choose_best")

    return plan
```

```python
# Step 3 — Execution Engine
def execute_plan(plan):
    context = {}

    for step in plan:
        if step == "search_flights":
            print("\n[Action] Searching Flights...")
            context["flights"] = search_flights("Delhi", "Mumbai")
            print(context["flights"])

        elif step == "check_weather":
            print("\n[Action] Checking Weather...")
            context["weather"] = check_weather("Mumbai")
            print(context["weather"])

        elif step == "choose_best":
            print("\n[Action] Choosing Best Option...")
            best = calculate_budget(context["flights"])
            context["best_flight"] = best
            print(best)

    return context
```

```python
# Step 4 — Full Agent
def travel_agent(user_query):
    print("User Query:", user_query)

    # Step 1: Plan
    plan = agent_brain(user_query)
    print("\n[Plan]", plan)

    # Step 2: Execute
    result = execute_plan(plan)

    # Step 3: Final Response
    final_output = f"""
Best Flight: {result['best_flight']['airline']} at ₹{result['best_flight']['price']}
Weather: {result['weather']['condition']}, {result['weather']['temperature']}°C
Recommendation: Book the cheapest early flight.
"""
    return final_output
```

```python
# Step 5 — Run It
response = travel_agent("Find best flight from Delhi to Mumbai")
print("\nFinal Answer:\n", response)
```

Scenario: Smart Shopping Assistant 🛒
You say to your students:
“Imagine you are planning a party and you tell your assistant:
‘Find me the best chocolate cake under ₹500.’
Now instead of just replying with text, your assistant actually thinks, checks different bakery websites, compares prices, and then gives you the final decision.”

```python
# ==========================================================
# 🛒 SMART SHOPPING AGENT (Agentic AI Example)
# ==========================================================



def search_cakes():
    """Simulates searching cakes from websites"""
    return [
        {"name": "Chocolate Truffle", "price": 450, "rating": 4.5},
        {"name": "Dark Fantasy Cake", "price": 550, "rating": 4.8},
        {"name": "Chocolate Cream Cake", "price": 400, "rating": 4.2},
        {"name": "Belgian Chocolate Cake", "price": 500, "rating": 4.7}
    ]


def filter_by_budget(cakes, budget):
    """Filters cakes within budget"""
    return [cake for cake in cakes if cake["price"] <= budget]


def choose_best_cake(cakes):
    """Chooses best cake based on rating"""
    if not cakes:
        return None

    best = max(cakes, key=lambda x: x["rating"])

    return {
        "cake": best,
        "reason": f"Selected '{best['name']}' because it has highest rating ({best['rating']}) within budget."
    }



def agent_brain(user_query):
    """Decides what steps to take"""
    plan = []

    if "cake" in user_query.lower():
        plan = [
            "search_cakes",
            "filter_budget",
            "choose_best"
        ]

    return plan



def execute_plan(plan, budget):
    context = {}

    for step in plan:

        if step == "search_cakes":
            print("\n [Action] Searching Cakes...")
            context["cakes"] = search_cakes()
            print(context["cakes"])

        elif step == "filter_budget":
            print("\n [Action] Filtering by Budget...")
            context["filtered"] = filter_by_budget(context["cakes"], budget)
            print(context["filtered"])

        elif step == "choose_best":
            print("\n [Action] Choosing Best Cake...")
            result = choose_best_cake(context["filtered"])
            context["best"] = result
            print(result)

    return context




def shopping_agent(user_query, budget):
    print(" User Query:", user_query)

    # 1. Plan
    plan = agent_brain(user_query)
    print("\n Plan:", plan)

    # 2. Execute
    result = execute_plan(plan, budget)

    # 3. Final Output
    if result["best"] is None:
        return " No cake found within your budget."

    cake = result["best"]["cake"]
    reason = result["best"]["reason"]

    final_output = f"""
 BEST CAKE FOUND
-------------------
Name: {cake['name']}
Price: {cake['price']}
Rating:  {cake['rating']}

🧠 Reason: {reason}

 Recommendation: Go for this cake!
"""
    return final_output




if __name__ == "__main__":
    query = "Find me the best chocolate cake"
    budget = 500

    result = shopping_agent(query, budget)
    print(result)
```

Scenario: The University’s AI Plagiarism Agent
Imagine a university has deployed an AI-powered plagiarism detection agent named PlagiarismLLM.
Its job is to monitor student essay submissions in real time and decide whether they are original or potentially plagiarized.

Step 1: Thinking (Reasoning)
Whenever an essay is submitted, PlagiarismLLM first analyzes the text:
- Example: “Essay contains large verbatim passages from online sources”
- PlagiarismLLM thinks: “Analyzing essay: multiple verbatim passages detected.”

Step 2: Planning (Choosing an Action)
Based on the thought, PlagiarismLLM decides which tool to use:
- If the essay has long verbatim matches or suspicious similarity scores, it plans to flag as potential plagiarism.
- Otherwise, it plans to mark as original work.

Step 3: Acting (Running Tools)
PlagiarismLLM hands over the decision to PlagiarismTools, which executes the action:
- flag_plagiarism → returns “Potential Plagiarism Detected”
- mark_original → returns “Original Work”

Step 4: Answering (Final Decision)
Once a clear classification is reached, PlagiarismLLM announces the decision:
- “Decision: Potential Plagiarism Detected”
- “Decision: Original Work”

```python
# ==========================================================
# 🎓 PLAGIARISM AI AGENT (Agentic AI Example)
# ==========================================================

# -------------------------------
# STEP 1 — TOOLS (Actions)
# -------------------------------

def flag_plagiarism():
    """Tool to flag plagiarism"""
    return " Potential Plagiarism Detected"


def mark_original():
    """Tool to mark as original"""
    return " Original Work"


# -------------------------------
# STEP 2 — THINKING (Reasoning)
# -------------------------------

def analyze_essay(essay_text):
    """
    Simulates analyzing essay content
    (In real-world → embeddings + similarity search)
    """

    suspicious_keywords = ["copy", "paste", "wikipedia", "source", "verbatim"]

    matches = [word for word in suspicious_keywords if word in essay_text.lower()]

    if matches:
        thought = f"Analyzing essay: suspicious patterns found ({matches})"
        risk_score = 0.8
    else:
        thought = "Analyzing essay: no major issues detected"
        risk_score = 0.2

    return thought, risk_score


# -------------------------------
# STEP 3 — PLANNING
# -------------------------------

def agent_planner(risk_score):
    """
    Decide which action to take
    """

    if risk_score > 0.5:
        plan = "flag_plagiarism"
    else:
        plan = "mark_original"

    return plan


# -------------------------------
# STEP 4 — ACTION (Execution)
# -------------------------------

def execute_action(plan):
    """
    Executes the chosen tool
    """

    if plan == "flag_plagiarism":
        print("\n [Action] Flagging Plagiarism...")
        return flag_plagiarism()

    elif plan == "mark_original":
        print("\n [Action] Marking as Original...")
        return mark_original()


# -------------------------------
# STEP 5 — FULL AGENT
# -------------------------------

def plagiarism_agent(essay_text):
    print(" Essay Submitted:\n", essay_text)

    # 1. THINK
    thought, risk_score = analyze_essay(essay_text)
    print("\n [Thinking]")
    print(thought)
    print(f"Risk Score: {risk_score}")

    # 2. PLAN
    plan = agent_planner(risk_score)
    print(f"Selected Action: {plan}")

    # 3. ACT
    result = execute_action(plan)

    # 4. FINAL DECISION
    final_output = f"\n Decision: {result}"

    return final_output


# -------------------------------
# STEP 6 — RUN EXAMPLES
# -------------------------------

if __name__ == "__main__":


    essay1 = "This essay was copy paste from wikipedia source."
    print(plagiarism_agent(essay1))

    essay2 = "This essay is written in my own words with original ideas."
    print(plagiarism_agent(essay2))
```

```python
import requests
from google.colab import userdata
API_TOKEN = userdata.get('hugging_api')

MODEL_ID = "meta-llama/Meta-Llama-3-8B-Instruct"

API_URL = "https://router.huggingface.co/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {API_TOKEN}",
    "Content-Type": "application/json"
}

def query_llm(prompt: str):
    payload = {
        "model": MODEL_ID,
        "messages": [
            {"role": "user", "content": prompt}
        ],
        "max_tokens": 200
    }

    response = requests.post(API_URL, headers=headers, json=payload)
    data = response.json()

    if "choices" in data:
        return data["choices"][0]["message"]["content"]
    else:
        return str(data)


def flag_suspicious(transaction):
    return "Suspicious Transaction"


def mark_normal(transaction):
    return "Normal Transaction"


def fraud_agent(transaction: str):
    thought = query_llm(f"Analyze this transaction: {transaction}")
    print("LLM Thought:", thought)

    if "overseas" in transaction.lower() or "multiple deposits" in transaction.lower():
        action = flag_suspicious
    else:
        action = mark_normal

    result = action(transaction)

    final_answer = query_llm(f"Decision: {result}")
    return final_answer


print(fraud_agent("Payment to overseas account"))
```

🏥 Scenario: The Hospital’s AI Appointment Agent
Imagine a hospital has deployed an AI-powered agent named MediLLM.
Its job is to monitor patient appointment requests and decide whether the requested slot is available or needs rescheduling.

Step 1 — Thinking (Reasoning)
Whenever a patient request comes in, MediLLM first analyzes the details:
- Example: “Request for Dr. Sharma at 10 AM tomorrow”
- MediLLM thinks: “Analyzing appointment request: Dr. Sharma, 10 AM tomorrow.”

Step 2 — Planning (Choosing an Action)
Based on the thought, MediLLM decides which tool to use:
- If the requested slot is free, it plans to confirm appointment.
- If the slot is already booked, it plans to suggest rescheduling.

Step 3 — Acting (Running Tools)
MediLLM hands over the decision to AppointmentTools, which executes the action:
- confirm_slot → returns “Appointment Confirmed”
- suggest_reschedule → returns “Slot Unavailable, Please Reschedule”

Step 4 — Answering (Final Decision)
Once a clear classification is reached, MediLLM announces the decision:
- “Decision: Appointment Confirmed”
- “Decision: Slot Unavailable, Please Reschedule”

```python
# -------------------------------
# MOCK DATABASE (Doctor Schedule)
# -------------------------------
appointments_db = {
    "Dr. Sharma": ["9 AM", "10 AM"],
    "Dr. Mehta": ["11 AM"]
}


def confirm_slot(doctor, time):
    return f"Appointment Confirmed with {doctor} at {time}"

def suggest_reschedule(doctor, time):
    return f"Slot Unavailable for {doctor} at {time}. Please Reschedule"



def extract_details(request):
    doctor = None
    time = None

    if "sharma" in request.lower():
        doctor = "Dr. Sharma"
    elif "mehta" in request.lower():
        doctor = "Dr. Mehta"

    if "10" in request:
        time = "10 AM"
    elif "9" in request:
        time = "9 AM"
    elif "11" in request:
        time = "11 AM"

    return doctor, time


def mediLLM_agent(request: str):



    doctor, time = extract_details(request)
    print(f"Analyzed: {doctor}, {time}")



    if time in appointments_db.get(doctor, []):
        action = suggest_reschedule
    else:
        action = confirm_slot

    result = action(doctor, time)


    return f"Decision: {result}"

print(mediLLM_agent("Request for Dr. Sharma at 10 AM tomorrow"))
```
