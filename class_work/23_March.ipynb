```python
# Scenario: AI Content Automation for a Startup
# 🏢 Problem Statement

# A startup wants to:

# Stay updated with latest AI trends

# Automatically create LinkedIn posts

# Save time on manual research & writing

# 👉 So we build a multi-agent AI system to automate this using CrewAI.

!pip install crewai litellm langchain-community langchain-litellm google-search-results

import os
os.environ["CREWAI_TRACING_ENABLED"] = "true"
from crewai import Agent, Task, Crew
from google.colab import userdata
from langchain_litellm import ChatLiteLLM # Updated import

# -----------------------------
# 1. Set Groq API Key
# -----------------------------
from google.colab import userdata
import os

# Fetch the secret from Colab (ensure the name matches your Secret key name)
api_key = userdata.get('api_key')

if api_key:
    os.environ["GROQ_API_KEY"] = api_key
else:
    print("❌ Error: 'GROQ_API_KEY' not found in Colab Secrets!")

# Double-check that OpenAI isn't interfering
os.environ.pop("OPENAI_API_KEY", None)
# -----------------------------
# 2. Define LLM (using ChatLiteLLM for Groq)
# -----------------------------
llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

# -----------------------------
# 3. Define Agents
# -----------------------------
researcher = Agent(
    role="Researcher",
    goal="Find latest AI news",
    backstory="Expert in AI trends",
    llm=llm, # Use the explicitly defined llm object
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Write LinkedIn post",
    backstory="Expert content creator",
    llm=llm, # Use the explicitly defined llm object
    verbose=True
)

# -----------------------------
# 4. Tasks
# -----------------------------
research_task = Task(
    description="Find 5 latest AI news trends",
    expected_output="List with summaries",
    agent=researcher
)

write_task = Task(
    description="Write LinkedIn post",
    expected_output="Final post",
    agent=writer,
    context=[research_task]
)

# -----------------------------
# 5. Crew
# -----------------------------
cryw = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    verbose=True
)

# -----------------------------
# 6. Run
# -----------------------------
result = cryw.kickoff()

print("\n✅ FINAL OUTPUT:\n")
print(result)
```

Scenario: Automated Learning Hub for an EdTech Company
🎓 Problem Statement
An EdTech company wants to:
- Keep track of the latest educational technology innovations
- Automatically generate weekly blog posts for their website
- Reduce time spent on manual research and content drafting
👉 So we build a multi-agent AI system using CrewAI to automate this workflow:
- Trend Monitoring Agent → Continuously scans for new EdTech tools, teaching strategies, and AI-in-education updates
- Content Summarization Agent → Condenses research into digestible insights tailored for educators and learners
- Blog Writing Agent → Crafts engaging, SEO-friendly blog posts automatically
- Publishing Agent → Schedules and posts content directly to the company’s website

```python
# =========================
# INSTALL (RUN ONCE)
# =========================
!pip install --upgrade crewai crewai-tools langchain-community

# =========================
# IMPORTS
# =========================
import os
from crewai import Agent, Task, Crew, Process, LLM
from crewai.tools import tool
from google.colab import userdata

# =========================
# API KEY
# =========================
os.environ["GROQ_API_KEY"] = userdata.get('api_key')

# =========================
# ROBUST LLM SETUP
# =========================
def get_llm():
    models = [
        "groq/llama-3.3-70b-versatile",
        "groq/llama-3.1-8b-instant",
    ]

    for model in models:
        try:
            print(f"Trying model: {model}")
            llm = LLM(
                model=model,
                api_key=os.environ["GROQ_API_KEY"]
            )
            print(f" Using model: {model}")
            return llm
        except Exception as e:
            print(f" Failed: {model} — {e}")

    raise Exception("No working model found. Check your Groq API key and model availability.")

llm = get_llm()

# =========================
# TOOL
# =========================
@tool("website_publisher")
def publish_to_website(title: str, content: str) -> str:
    """Publishes blog content to the website CMS."""
    print(f"\n🚀 Publishing Blog: {title}")
    return " Blog published successfully (draft mode)."

# =========================
# AGENTS
# =========================
trend_monitor = Agent(
    role="Trend Monitoring Agent",
    goal="Identify important EdTech innovations.",
    backstory="Expert in education technology trends.",
    verbose=True,
    llm=llm
)

summarizer = Agent(
    role="Summarization Agent",
    goal="Convert research into simple insights.",
    backstory="Simplifies complex ideas.",
    verbose=True,
    llm=llm
)

writer = Agent(
    role="Blog Writer",
    goal="Write engaging blog posts.",
    backstory="SEO content expert.",
    verbose=True,
    llm=llm
)

publisher = Agent(
    role="Publisher",
    goal="Publish blog content.",
    backstory="Handles CMS publishing.",
    tools=[publish_to_website],
    verbose=True,
    llm=llm
)

# =========================
# TASKS
# =========================
research_task = Task(
    description="List 3 important EdTech innovations and explain them.",
    expected_output="3 tools with benefits.",
    agent=trend_monitor
)

summary_task = Task(
    description="Summarize into 5 key points for teachers.",
    expected_output="5 bullet points.",
    agent=summarizer,
    context=[research_task]
)

blog_task = Task(
    description="Write a 500-word blog with title, intro, 3 sections, conclusion.",
    expected_output="Markdown blog post.",
    agent=writer,
    context=[summary_task]
)

publish_task = Task(
    description="Publish the blog using the website_publisher tool.",
    expected_output="Confirmation message.",
    agent=publisher,
    context=[blog_task]
)

# =========================
# CREW
# =========================
crew = Crew(
    agents=[trend_monitor, summarizer, writer, publisher],
    tasks=[research_task, summary_task, blog_task, publish_task],
    process=Process.sequential,
    verbose=True
)

# =========================
# RUN
# =========================
print("🚀 STARTING SYSTEM...\n")
result = crew.kickoff()

print("\n" + "="*50)
print("FINAL OUTPUT:")
print("="*50)
print(result)
```

[12:16, 3/23/2026] Himanshu Sardana: Scenario: Smart Sales Proposal Assistant
📌 Problem Statement
Enterprise sales teams often struggle to:
- Analyze client requirements from RFPs (Requests for Proposals)
- Match company offerings to client needs
- Draft tailored proposals that stand out
- Ensure consistency across documents
👉 So we build a multi-agent AI system that automates sales proposal preparation.

🤖 System Overview
- 🔹 Agent 1: RFP Analyzer
- Reads client’s request for proposal
- Extracts key requirements, budget constraints, and evaluation criteria
- 🔹 Agent 2: Solution Mapper
- Matches company products/services to client needs
- Suggests tailored value propositions and differentiators
- 🔹 Agent 3: Proposal Writer
- Generates a polished, client-specific proposal
- Ensures alignment with brand tone and compliance standards

🔄 Workflow
RFP Document → Requirement Analysis → Solution Mapping → Proposal Generation

```python
# =========================
# INSTALL (RUN ONCE)
# =========================
!pip install --upgrade crewai crewai-tools langchain-community

# =========================
# IMPORTS
# =========================
import os
from crewai import Agent, Task, Crew, Process, LLM
from crewai.tools import tool
from google.colab import userdata

# =========================
# API KEY
# =========================
os.environ["GROQ_API_KEY"] = userdata.get('api_key')

# =========================
# LLM SETUP
# =========================
def get_llm():
    models = [
        "groq/llama-3.3-70b-versatile",
        "groq/llama-3.1-8b-instant",
    ]
    for model in models:
        try:
            print(f"Trying model: {model}")
            llm = LLM(model=model, api_key=os.environ["GROQ_API_KEY"])
            print(f"✅ Using model: {model}")
            return llm
        except Exception as e:
            print(f"❌ Failed: {model} — {e}")
    raise Exception("No working model found.")

llm = get_llm()

# =========================
# SAMPLE RFP DOCUMENT
# =========================
rfp_document = """
CLIENT: GlobalRetail Inc.
PROJECT: Digital Transformation Initiative

REQUIREMENTS:
1. Cloud-based ERP system for 500+ users
2. Real-time inventory management across 20 warehouses
3. AI-powered demand forecasting
4. Integration with existing SAP and Salesforce systems
5. Mobile app for field sales team

BUDGET: $500,000 - $750,000
TIMELINE: 12 months implementation
EVALUATION CRITERIA: Technical capability (40%), Cost (30%), Support (20%), Innovation (10%)

COMPLIANCE: Must meet SOC2 Type II and GDPR requirements
PREFERRED VENDOR: Experience in retail sector preferred
"""

# =========================
# COMPANY OFFERINGS
# =========================
company_offerings = """
OUR COMPANY: TechSolutions Pro

PRODUCTS & SERVICES:
1. CloudERP Suite — cloud-based ERP for 100-10,000 users, $200K-$600K
2. WarehouseIQ — real-time inventory management, supports up to 50 locations
3. DemandAI — ML-powered demand forecasting, 94% accuracy
4. IntegrationHub — pre-built connectors for SAP, Salesforce, Oracle
5. MobileForce — mobile app for sales teams (iOS & Android)
6. 24/7 enterprise support with dedicated account manager

DIFFERENTIATORS:
- 15+ years in retail sector
- SOC2 Type II and GDPR certified
- Average implementation time: 10 months
- 200+ enterprise clients globally
- 99.9% uptime SLA
"""

# =========================
# TOOLS (only 2 — proposal_saver REMOVED to avoid Groq tool_use_failed bug)
# =========================
@tool("rfp_reader")
def read_rfp(query: str) -> str:
    """Reads and returns the RFP document content for analysis."""
    return rfp_document

@tool("company_catalog")
def get_company_offerings(query: str) -> str:
    """Returns company products, services, and differentiators."""
    return company_offerings

# =========================
# AGENTS
# =========================
rfp_analyzer = Agent(
    role="RFP Analyzer",
    goal="Extract and structure all key requirements, constraints, and evaluation criteria from the RFP document.",
    backstory="""You are a senior business analyst with 10+ years of experience reading
    enterprise RFPs. You extract hidden requirements, budget constraints, compliance needs,
    and decision criteria that determine whether a proposal wins or loses.""",
    tools=[read_rfp],
    verbose=True,
    llm=llm
)

solution_mapper = Agent(
    role="Solution Mapper",
    goal="Match company products and services to client requirements and craft compelling value propositions.",
    backstory="""You are a solutions architect who finds perfect matches between client needs
    and company offerings, and frames differentiators that win deals.""",
    tools=[get_company_offerings],
    verbose=True,
    llm=llm
)

proposal_writer = Agent(
    role="Proposal Writer",
    goal="Write a polished, persuasive, client-specific sales proposal that wins the deal.",
    backstory="""You are an award-winning proposal writer who has helped close $50M+ in enterprise
    deals. You write with clarity, confidence, and client focus.""",
    tools=[],   # ✅ No tools — avoids Groq tool_use_failed on large content
    verbose=True,
    llm=llm
)

# =========================
# TASKS
# =========================
analysis_task = Task(
    description="""
    Use the rfp_reader tool to read the RFP document.
    Extract and structure:
    1. Client name and project title
    2. All functional and technical requirements (numbered list)
    3. Budget range and timeline
    4. Evaluation criteria with weightages
    5. Compliance and regulatory requirements
    6. Any implicit needs or red flags
    """,
    expected_output="Structured analysis: client_info, requirements, budget, timeline, evaluation_criteria, compliance, implicit_needs",
    agent=rfp_analyzer
)

mapping_task = Task(
    description="""
    Use the company_catalog tool to get our offerings.
    Based on the RFP analysis:
    1. Map EACH requirement to specific company products/services
    2. Identify any gaps and how to address them
    3. Create 3-5 value propositions tailored to this client
    4. List top 5 competitive differentiators
    5. Suggest pricing within their budget
    6. Highlight compliance certifications
    """,
    expected_output="Solution mapping: requirement_coverage, value_propositions, differentiators, pricing_recommendation, compliance_alignment",
    agent=solution_mapper,
    context=[analysis_task]
)

proposal_task = Task(
    description="""
    Write a complete sales proposal using the analysis and mapping from previous tasks.

    Structure:
    1. Executive Summary (2 paragraphs)
    2. Understanding of Client Needs
    3. Proposed Solution (map to each requirement)
    4. Why Choose Us (differentiators + social proof)
    5. Implementation Timeline (month-by-month, 12 months)
    6. Investment and ROI (pricing within budget)
    7. Compliance and Security (SOC2, GDPR)
    8. Next Steps (clear call to action)

    Tone: Professional, confident, client-focused.
    Length: 600-800 words.
    Format: Markdown.

    Return ONLY the proposal text. Do not use any tools.
    """,
    expected_output="A complete polished markdown sales proposal (600-800 words).",
    agent=proposal_writer,
    context=[analysis_task, mapping_task]
)

# =========================
# CREW
# =========================
crew = Crew(
    agents=[rfp_analyzer, solution_mapper, proposal_writer],
    tasks=[analysis_task, mapping_task, proposal_task],
    process=Process.sequential,
    verbose=True
)

# =========================
# RUN
# =========================
print("🚀 STARTING SMART SALES PROPOSAL ASSISTANT...\n")
print("="*60)

result = crew.kickoff()

# =========================
# SAVE OUTPUT IN PYTHON (not via tool)
# =========================
proposal_text = str(result)

with open("sales_proposal.md", "w") as f:
    f.write("# Sales Proposal — GlobalRetail Inc.\n\n")
    f.write(proposal_text)

print("\n" + "="*60)
print("✅ PROPOSAL GENERATION COMPLETE")
print("="*60)
print("\n💾 Proposal saved to sales_proposal.md")
print("\n📄 PROPOSAL PREVIEW:")
print("-"*60)
print(proposal_text)
```
