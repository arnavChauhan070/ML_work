Scenario: Automated Business Report Assistant
🎯 Problem Statement
In companies:
- Managers spend hours compiling data from multiple sources (CRM, ERP, spreadsheets).
- Reports often require manual formatting, chart creation, and consistency checks.
- This slows down decision-making and introduces human error.
👉 Solution
A multi-agent AutoGen system where:
- 🤖 Assistant Agent drafts report templates and queries relevant data.
- ⚙️ UserProxy Agent executes data pulls and formatting automatically.
- 🔁 The system iterates until the report meets formatting, accuracy, and completeness standards.
🔄 Workflow
Business Request → AI drafts report structure → Data fetched & formatted → Charts auto-generated → Validation loop → Final polished report

```python
import os
import autogen
from google.colab import userdata

# 🔑 Load API Key
api_key = userdata.get("groq_api")

# Safety check
if not api_key or not api_key.startswith("gsk_"):
    raise ValueError("❌ Invalid Groq API Key")

os.environ["GROQ_API_KEY"] = api_key

# 📁 Working directory
os.makedirs("coding", exist_ok=True)

# 🤖 LLM Config
config_list = [
    {
        "model": "llama-3.3-70b-versatile",
        "api_key": api_key,
        "base_url": "https://api.groq.com/openai/v1"
    }
]

llm_config = {"config_list": config_list, "temperature": 0}

# 🧠 Assistant Agent (Report Generator)
assistant = autogen.AssistantAgent(
    name="Report_Generator",
    llm_config=llm_config,
    system_message="""
You are a business analyst AI.

Your job:
1. Generate structured reports
2. Write Python code to:
   - Analyze data
   - Create charts
3. Explain insights clearly

Always:
- Use pandas & matplotlib
- Print clean outputs
"""
)

# 🔍 Reviewer Agent (Validation Loop)
reviewer = autogen.AssistantAgent(
    name="Reviewer",
    llm_config=llm_config,
    system_message="""
You are a strict reviewer.

Check:
- Data correctness
- Completeness
- Business insights quality

If something is missing → ask for improvement.
If good → say 'APPROVED'
"""
)

# ⚙️ Executor Agent (runs code)
user_proxy = autogen.UserProxyAgent(
    name="Executor",
    human_input_mode="NEVER",
    max_consecutive_auto_reply=10,
    code_execution_config={
        "work_dir": "coding",
        "use_docker": False
    }
)

# 🔄 Group Chat (multi-agent system)
groupchat = autogen.GroupChat(
    agents=[assistant, reviewer, user_proxy],
    messages=[],
    max_round=10
)

manager = autogen.GroupChatManager(
    groupchat=groupchat,
    llm_config=llm_config
)

# 🚀 BUSINESS TASK
user_proxy.initiate_chat(
    manager,
    message="""
Create a business report for sales data:

Data: [12000, 15000, 10000, 18000, 13000]

Requirements:
1. Summary statistics (mean, median, min, max)
2. Generate a bar chart
3. Identify trends
4. Provide business insights
5. Ensure report is clean and complete

Reviewer must approve the final output.
"""
)
```

```python
# Scenario: AI Code Review & Execution System (Corporate Dev Team)
# 🎯 Problem Statement

# In companies:

# Developers write code

# Reviewers check for bugs

# Testing is manual

# 👉 Solution:
# A multi-agent system where:

# 👨‍💻 Coder Agent writes code

# 🔍 Reviewer Agent validates + executes code

# 🔁 Loop continues until correct


# -----------------------------
# Imports
# -----------------------------
import os
import autogen
from google.colab import userdata

# -----------------------------
# Groq API Key
# -----------------------------
groq_api_key = userdata.get("groq_api")
os.environ["GROQ_API_KEY"] = groq_api_key

# -----------------------------
# LLM Config (Groq)
# -----------------------------
llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": groq_api_key,
            "base_url": "https://api.groq.com/openai/v1"
        }
    ],
    "temperature": 0
}

# -----------------------------
# Agent 1: Coder
# -----------------------------
coder = autogen.AssistantAgent(
    name="Coder",
    llm_config=llm_config,
    system_message="""You are an expert Python developer.

Write clean, well-commented code.

Rules:
- Handle edge cases
- Add test cases
- Ensure code runs without errors
- End your response with DONE"""
)

# -----------------------------
# Agent 2: Reviewer (Executor)
# -----------------------------
reviewer = autogen.UserProxyAgent(
    name="Reviewer",
    human_input_mode="NEVER",
    code_execution_config={
        "work_dir": "review",
        "use_docker": False
    },
    is_termination_msg=lambda x: "DONE" in x.get("content", ""),
    system_message="""You are a code reviewer.

Responsibilities:
1. Review the code for bugs
2. Execute the code
3. Validate outputs
4. Ask for fixes if incorrect
5. Approve when correct"""
)

# -----------------------------
# Start Collaboration
# -----------------------------
reviewer.initiate_chat(
    coder,
    message="""Create a Python function that:
- Generates Fibonacci sequence up to N
- Returns it as a list
- Handles edge cases (N=0, negative N)

Test it with N=10 and print output"""
)
```
